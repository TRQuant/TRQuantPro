```python
# 直接使用 graphviz - 最简方案，无需任何自定义库
import graphviz

# Dark Mode 颜色
BG = '#1E1E1E'
EDGE = '#BDBDBD'
SUCCESS = '#2E7D32'
PROCESS = '#1565C0'
RESEARCH = '#6A1B9A'
DATA = '#00838F'

dot = graphviz.Digraph(format='svg')
dot.attr(rankdir='TB', splines='spline', bgcolor=BG, label='测试流程图', fontcolor='white', labelloc='t')
dot.attr('node', shape='box', style='rounded,filled', fontname='SimHei', fontcolor='white', fontsize='11')
dot.attr('edge', color=EDGE, penwidth='1.5')

dot.node('A', '开始', fillcolor=SUCCESS)
dot.node('B', '数据获取', fillcolor=DATA)
dot.node('C', '市场分析', fillcolor=RESEARCH)
dot.node('D', '策略生成', fillcolor=PROCESS)
dot.node('E', '结束', fillcolor=SUCCESS)

dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')

dot
```

```python
# 横向流程图
dot2 = graphviz.Digraph(format='svg')
dot2.attr(rankdir='LR', splines='spline', bgcolor=BG, label='投资工作流程', fontcolor='white', labelloc='t')
dot2.attr('node', shape='box', style='rounded,filled', fontname='SimHei', fontcolor='white')
dot2.attr('edge', color=EDGE, penwidth='1.5')

dot2.node('D1', '数据源', fillcolor=DATA)
dot2.node('R1', '研究阶段', fillcolor=RESEARCH)
dot2.node('L1', '实战阶段', fillcolor='#D84315')

dot2.edge('D1', 'R1')
dot2.edge('R1', 'L1')

dot2
```
